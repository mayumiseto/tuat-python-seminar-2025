# 2-1. リスト/配列の基本

- Pythonでは、複数の値をまとめて扱うには **リスト** を使う。
- 角かっこ `[]` で囲み、カンマで区切る。

````python
# 1週間分の日平均気温データ（サンプル）
temps_tokyo = [27.4, 27.1, 28.0, 29.3, 28.8, 27.9, 28.2]
print("temps_tokyo のデータ数は", len(temps_tokyo))
print(temps_tokyo)
````

- インデックス（番号）で特定の要素にアクセスできる。
- ⚠️ リストの先頭は1番目ではなく0番目のインデックスが割り当てられていることに注意。

````python
print(temps_tokyo[0])  # 最初の要素（27.4）
print(temps_tokyo[6]) # 最後の要素（28.2）
print(temps_tokyo[-1]) # 最後の要素（28.2）
````


- 表データ（列がそろった表形式のデータ）は **pandas** の **DataFrame** で扱うのが便利。
- **Series** は「1列のデータ（縦ベクトル）」、**DataFrame** は「複数列を束ねた表」だと思えばOK。

### 最小例：リストからDataFrameを作る
下の例では、日付のリスト `dates` と、対応する気温のリスト `temps_tokyo` を 1 つの DataFrame に束ねる。


````python

from datetime import date, timedelta
import pandas as pd

# サンプル：1980年8月1日〜7日までの日平均気温
temps_tokyo = [21.6, 20.5, 19.5, 20.5, 21.8, 21.4, 22.2]

# 日付のリストを作成（今日から過去6日分）
start = date(1980, 8, 1)  # 固定日付でもOK
dates = [start + timedelta(days=i) for i in range(len(temps_tokyo))]

# DataFrameを作成
df = pd.DataFrame({
    "date": dates,
    "temp_C": temps_tokyo
})

df.head()
````


### DataFrameの基本操作
- 先頭/末尾の確認：`df.head()`, `df.tail()`
- 列の取り出し：`df["temp_C"]`
- 要素数：`len(df)`、平均：`df["temp_C"].mean()`
- 条件で抽出：`df[df["temp_C"] >= 30]`


````python

print("行数:", len(df))
print("平均(℃):", df["temp_C"].mean())

# 30℃以上の日だけ表示
hot = df[df["temp_C"] >= 30]
hot
````

# 2-2. 繰り返し処理(for文)


## for文の構文（Python）
- **所定回数の繰り返し**には `for` と `range(...)` を組み合わせる。

```python
for i in range(N):
    # i は 0, 1, 2, ..., N-1 と変化
    # ここに繰り返したい処理
```

- **リストの全要素に順にアクセス**する場合：

```python
for x in some_list:
    # x には各要素が順に入る
```

- **インデックスと値を同時に使いたい**場合は `enumerate`：

```python
for idx, x in enumerate(some_list):
    # idx は 0,1,2..., x は値
```


````python

temps_tokyo = [27.4, 28.1, 29.0, 30.2, 31.3, 29.8, 28.5]

# 例1: 繰り返し回数を決めて出力
for i in range(3):
    print("Hello", i)

# 例2: リストの値を合計（手書きfor）
s = 0.0
for t in temps_tokyo:
    s += t
print("合計(手計算):", s)

# 例3: enumerateでインデックスと値を同時に
for i, t in enumerate(temps_tokyo):
    print(f"{i}日目: {t} ℃")
````


### 手書きのfor と 組み込み関数の違い
- **学習初期**は「forで中身を自分で計算する」ことで計算手順を理解する。
- **実務**では、`sum`, `min`, `max`, `len`, `statistics.mean` や、`pandas` のメソッドなど **既存の道具**を優先する。
- ただし、**関数名で変数を上書き**してしまうとエラーの原因になるので注意（例：`sum = 0` としてしまう など）。


- range(5)は 0 から < 5 の整数の並びを生成する関数。

````python
for i in range(5):
    print(i)
````

````python
for i in range(len(temps_tokyo)):
    print(temps_tokyo[i])
````

````python
for temp in temps_tokyo:
    print("気温, ", temp)
````

````python
for temp in temps_tokyo:
    print("気温, ", temp)
````

| 構文要素       | 説明                         |
| ---------- | -------------------------- |
| `for`      | 「～に対して順に処理を行う」ことを示すキーワード   |
| `temp`     | ループの中で使われる一時的な変数（1つずつ値が入る） |
| `in`       | 「データ群の中の」という意味を表す          |
| `temps`    | リスト（データの集まり）               |
| `:`        | この下の行が繰り返し処理の中身であることを示す    |
| インデント   | インデントから始まる行が繰り返し処理の中身であることを示す    |


# 2-3. 繰り返し処理と組み込み関数

- まずは1週間分の日平均気温データから、for文を用いて1週間の平均気温を計算する方法を考えてみよう。
- for文の中で気温データの総和を計算し、最後に日数で割ることで平均を得る。

````python
sum = 0
for tmp in temps_tokyo:
    sum = sum + tmp
print(f"総和: {sum}")
print(f"平均: {sum/7:.2f}")
````

````python
- 実は、sum(list)関数というlistの総和を返す関数もある。自分でfor文を用いて計算した結果と比較してみよう。
````

````python
sum(temps_tokyo)
````

- 上記を実行するとエラーが出るはずだ。このエラーは一個上のセルでPythonにもともと存在していた組み込み関数sumに数値を上書きしてしまったことが原因である。このように、気付かないうちに変数や関数に上書きしてしまってエラーが生じることはしばしば発生する。こうしたときは、一度変数や関数をprintで表示してみるのが定石。

````python
print(sum)
````

- 上書きされてしまったsumを解放し、組み込み関数に戻すためには、以下のように削除を行う。

````python
del sum
````

- もう一度printで表示してみよう。<built-in function sum> (組み込み関数 sum) が表示されるはず。もう一度sum関数による総和の計算を試してみよう。

````python
print(sum)
sum(temps_tokyo)
````

- 以下にいくつかの組み込み関数を紹介し、その組み込み関数の働きをfor文で実装する例を紹介する。

| 関数         | 説明        |
| ---------- | --------- |
| `sum(リスト)` | すべての値の合計  |
| `len(リスト)` | データ数（要素数） |
| `max(リスト)` | 最大値       |
| `min(リスト)` | 最小値       |


````python
# max関数による最大値の取得
maxv = max(temps_tokyo)
print(f"最大値: {maxv:.2f}")

# for文による最大値の取得の実装
tmp = temps_tokyo[0]
for temp in temps_tokyo:
    if tmp < temp:
        tmp = temp
print(f"最大値: {tmp:.2f}")
````


### ループで平均を計算 → pandas で平均を計算
同じ平均でも、段階に応じて **(1) forで手計算** → **(2) `sum` / `len` で簡潔** → **(3) pandas で本格的** と使い分けよう。


````python

# (1) forで手計算
s = 0.0
for v in df["temp_C"]:
    s += v
avg1 = s / len(df)

# (2) sum/len
avg2 = sum(df["temp_C"]) / len(df)

# (3) pandas
avg3 = df["temp_C"].mean()

avg1, avg2, avg3
````
