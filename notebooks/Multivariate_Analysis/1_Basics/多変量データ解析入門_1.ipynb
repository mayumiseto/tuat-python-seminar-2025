## このノートで学ぶこと

```python
import pandas as pd
import numpy as np
```

## 1. DataFrameとは?
- Excelの表のような「行×列」構造をもつ2次元データ。
- pandas における Series（1列ベクトル）を複数まとめたものが DataFrame。

```python
# 例: シンプルな表を作る

data = {
    'Name': ['John', 'Mark', 'Emily', 'Criss'],
    'Age': [18, 19, 20, 21],
    'Score': [85, 92, 78, 88]
}
df = pd.DataFrame(data)
df.index = ['Student 1', 'Student 2', 'Student 3', 'Student 4']
print(df)
```

```python
# DataFrame の構造要素
print('\n--- 行と列の情報 ---')
print('列名 columns:', df.columns.tolist())
print('行番号 index:', df.index.tolist())
print('形状 shape:', df.shape)
```

## 2. loc と iloc の違い
- DataFrameの中身を取り出すときの最重要メソッド。
- loc  → ラベル（index名や列名）で指定する。
- iloc → 数値インデックス（位置）で指定する。

```python
# 例: 2行目のデータを取り出す
print('\n--- loc（ラベル指定）---')
print(df.loc['Student 1'])  # 行ラベル1のデータ

print('\n--- iloc（位置指定）---')
print(df.iloc[1])  # 2行目（0始まりなので位置は1になることに注意）

# 例: 複数列を指定
print('\n--- 列の選択 ---')
print(df.loc[:, ['Name', 'Score']])  # 全行から特定列を選択

# 例: 部分的な範囲指定
print('\n--- 範囲指定 ---')
print(df.iloc[1:3, 0:2])  # 2〜3行目, 1〜2列目
```

## 3. loc と iloc の応用（データ抽出・条件選択）

```python
# 条件を使った抽出（locを使う）
print('\n--- 条件抽出: Score>=85 ---')
print(df.loc[df['Score'] >= 85])

# 位置を使った柔軟抽出（iloc）
print('\n--- ilocを使って先頭2行を取得 ---')
print(df.iloc[:2])
```

## 4. DataFrameの中身を理解する時に便利な関数
1) df.head() → 先頭5行を確認（データの形の把握）
2) df.info() → 各列の型や欠損の有無を確認
3) df.describe() → 数値列の統計を一括確認

```python
print('df.head():')
print(df.head())
print('\ninfo():')
print(df.info())
print('\ndescribe():')
print(df.describe())
```
