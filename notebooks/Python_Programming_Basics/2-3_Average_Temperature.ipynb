## Step 3-1：テキストファイルの読み書き（open / read / write）

````python
f = open("sample.txt", "r")
text = f.read()
f.close()

print(text)
````

読み込みのコードの流れは以下の通り。

1. open でファイルを開く
2. read() で中身を読む
3. close() でファイルを閉じる

- **open**: ファイルを開くための関数
    Python では、ファイルを読み書きする前に、必ず `open` を使って「このファイルを使います」と宣言する必要がある。

```python
f = open("ファイル名", "モード")
```
    
- **"r"**:「読み込み（read）」モード
- **read**: ファイルの中身を全部読むための関数。
- **close()**: ファイルを閉じるための関数。

【備考】

ファイルを開いたままにすると、

- データが正しく保存されない
- 他のプログラムがファイルを使えない

といった問題が起こることがあるため、使い終わったら必ず close() する。

ファイルを書き出す流れは、以下の通り。

1. open("ファイル名", "w")でファイルを開く (w は書き込み（write）モード）
2. write()で書き込む
3. close()でファイルを閉じる

````python
f = open("output.txt", "w")
f.write("Hello, Python!")
f.close()
````

````python
f = open("output.txt", "r")
text = f.read()
f.close()

print(text)
````

## Step 3-2:CSV ファイル（表データ）を Python で扱う

### CSVとは？

- CSV（Comma Separated Values）は、カンマ（,）で区切られている表形式のデータである。
- Excel でも開けるため、データ解析においてしばしば利用されるデータ形式の1つ。


以下は tokyo_daily_mean_temperature_aug_1980.csv の CSV ファイルを1行ずつ読み込み、その中身を f として読み込み、1行ずつ中身を表示する。

````python
#csvは、CSVファイル（表データ）を簡単に扱うための標準ライブラリであり、 Python には最初から入っているため、インストール不要。
import csv

with open("tokyo_daily_mean_temperature_aug_1980.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

- この CSV ファイルの中身は、

| 列 | 内容 |
|---|---|
| 1列目 | 日付（yyyy-08-dd） |
| 2列目 | 東京の日平均気温（℃） |

### with open(...) as f: の意味

```python
with open("tokyo_daily_mean_temperature_aug_1980", "r", encoding="utf-8") as f:
```

- with 文を使うと、

    1. ファイルを開く
    2. 処理を行う
    3. 処理が終わったら 自動的にファイルを閉じる

    という流れを、close()無しにで実行できる。

- as f は、「開いたファイルを f という名前で用いる」の意味。


### encoding とは何か？

`encoding`（エンコーディング）とは、  
文字をコンピュータがどのような数字として保存・読み取るかのルール。

私たちが見ている

- 日本語（例：東京、気温）
- 英語（例：temperature）
- 記号（例：℃、％）

は、コンピュータの中ではすべて 数値の列として保存されている。  
その「数値と文字の対応表」が 文字コード（encoding）であり、
適切な対応表を指定しないと文字化けが起こる:

```text
東京 → æ�±äº¬
```

### 日本語でよく使われる文字コード

| 文字コード | 主な使用場面 | 特徴 |
|---|---|---|
| UTF-8 | 現在の標準 | 世界共通・最も安全 |
| UTF-8-SIG | Excel用UTF-8 | Excelで文字化けしにくい |
| Shift_JIS | 日本の古い環境 | Windows・Excelで多用 |
| CP932 | Shift_JISの拡張 | 日本語Windows内部 |

### csv.reader(f)の役割
- f（開いたCSVファイル）を
- csv.reader に渡して
- 1行ずつ読み取れる形に変換する

### for row in reader: の意味
- CSV ファイルの1行ずつを順番に取り出して処理する。下記の場合は一行ずつ表示。

```python
for row in reader:
    print(row)
```

以下では、CSVファイルに保存された1980年の東京の8月の日平均気温データを1行ずつ読み込み、

- 日付
- 気温

をそれぞれ別のリストに保存する。

````python
dates=[] #空のリストを作成する
temps=[] #空のリストを作成する
with open("tokyo_daily_mean_temperature_aug_1980.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader) # CSVの1行目（列名）を読み取り、データ処理から除外する。以降readerは元々のcsvファイルの2行目以降のみを対象とする。

    for row in reader:
        date = row[0] #1列目をdateに代入
        temp = row[1] #2列目をtempに代入
        dates.append(date)  #dateをdatesに追加
        temps.append(float(temp))  #tempをfloat型に変換後、をtempsに追加
````

````python
print("Length of temps:", len(temps))
print("Length of dates:", len(dates))
print(dates)
print(temps)
````

tempsを用い、1980年の東京の8月の平均気温を計算する。

````python
mean = sum(temps) / len(temps)
print(f"Mean: {mean:.2f}")
````

## 3-3. 条件抽出

以下は tokyo_daily_mean_temperature_aug_1980_2025.csv の CSV ファイルを1行ずつ読み込む。このファイルには、1980年から2025年までの東京の8月の日平均気温が保存されている。このデータから**yyyy年**の条件で日平均気温を抽出し、その年の8月の月平均を計算する。

````python
import csv

dates = []
temps = []

with open("tokyo_daily_mean_temperature_aug_1980_2025.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)  # 列名を読み飛ばす

    for row in reader:
        date = row[0]   # 例: "1980-08-01"
        temp = row[1]
        dates.append(date)
        temps.append(float(temp))
````

### (i) dates と temps を「2列データ」にする

pandas の DataFrame を使い、dates と temps の2列データを作成する。

````python
import pandas as pd

df = pd.DataFrame({
    "date": dates,
    "temp": temps
})
````

````python
print(df)
````

### (ii) 日付を「日付型」に変換する

文字列のままだと月ごとの抽出や並び替えがしにくいので、to_datetime を使うことで、日付の扱いを容易にする。

````python
import pandas as pd
df["date"] = pd.to_datetime(df["date"])
````

### (iii) 年だけを抽出し、新たな列として作成

年で条件抽出や平均を取るため、year 列を新しく作る。**df["date"].dt.year**が年の抽出であり、これを**df["year"]**に代入することで"year"の列名の列が追加される。

````python
df["date"].dt.year
````

````python
df["year"] = df["date"].dt.year
````

````python
print(df)
````

以下は列とそのデータ型の確認

````python
df.columns
````

````python
df.dtypes
````

### (iv) 特定の年だけを条件抽出する

以下では、this_yearに"year"列が2025と一致する行のみが代入される。this_yearの"temp"列に対して平均を求めることで、2025年の8月の月平均気温を計算できる。

````python
this_year = df[df["year"]==2025]
````

````python
this_year_temp = this_year["temp"]
this_year_mean = sum(this_year_temp) / len(this_year_temp)
print(f"Mean: {this_year_mean:.2f}")
````

### (v) 1980, 1981, ..., 2025の数値列を作成し、forループで各年の8月平均気温を取得

````python
years = range(1980, 2026)
year_means = []

for y in years:
    mean_temp = df[df["year"] == y]["temp"]
    year_means.append(sum(mean_temp)/len(mean_temp))
````

### (vi) years と years_means で再度データフレームを作成

````python
df2 = pd.DataFrame({
    "year": years,
    "ave_temp": year_means
})
````

````python
print(df2)
````

## 3-5. 計算結果を新しい CSV ファイルとして保存する

上記で作成した df2 をCSVファイルとして保存

````python
df2.to_csv(
    "tokyo_mean_aug_1980_2025.csv",
    index=False,        # 行番号（index）を出力しない
    encoding="utf-8"    # 文字コードを UTF-8 に指定
)
````
