# Jupyter Notebook とは？

- Jupyter Notebook（ジュピター・ノートブック）は、**プログラムコード・説明文・実行結果を1つの画面でまとめて扱える**ツール。
- Pythonの学習やデータ解析、研究記録に広く使われる。

---

## セル（cell）とは？

Notebookは「セル」と呼ばれるブロックで構成され、主に3種類がある。

| セルの種類 | 役割 | 使用例 |
|-------------|------|--------|
| **Code（コード）** | Pythonのプログラムを書く場所。<br>Shift + Enter で実行できる。 | `print("Hello, world!")` |
| **Markdown（マークダウン）** | 説明文・数式・見出しを書くセル。<br>HTMLのように整形できる。 | `# タイトル`、`**強調**`、`_斜体_` |
| **Raw（ロー）** | そのまま出力されるテキスト。通常は使わない。 | 出力用メモやLaTeX原稿など |

---

## Codeセルの使い方

1. セルの左側に「`In [ ]:`」と表示されているのがCodeセル。  
2. 中にPythonコードを書く。  
3. **Shift + Enter** で実行。  
4. 実行結果はすぐ下に表示される。


```python
print("Hello World")
```

## 出力：print() 関数

- プログラムの結果を確認するためには、出力（print）が必要。
- Pythonでは `print()` 関数を使って、変数や文字列の内容を画面に表示する。
