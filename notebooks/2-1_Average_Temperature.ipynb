# 1-1. 変数とデータ型

## 変数とは？
- プログラム中で値（データ）を一時的に保存したり、計算に利用するために「変数（variable）」を使う。  
変数は「名前のついたデータの入れ物」と考えることができる。
- 変数に値を格納するには =（代入演算子）を使う。
- 以下ではtemperatureという変数に23.8という浮動小数点型の値を、locationという変数に文字列を、それぞれ格納する。

```python
temperature = 23.8   # 水温（℃）
location = "Lake Biwa"  # 地点名
```

## 変数のデータ型

- 変数には種類（データ型, type）がある。
- 水温なら小数を扱う**浮動小数点数（float）**、地点名なら文字列（string, `str`）。
- Pythonでは型を自動で判断してくれるため、C言語やJavaのように `int x;` のような宣言は不要。

---

## 型の例

| データの種類 | 型の名前 | 例 |
|---------------|------------|----|
| 整数 | `int` | 10, -5, 2025 |
| 実数（小数） | `float` | 23.8, -1.2, 3.14 |
| 文字列 | `str` | "Lake Biwa", 'Tokyo' |
| 論理値 | `bool` | True, False |

※ Pythonでは文字列を表すのにシングルクオテーション('')とダブルクオテーション("")の両方を用いる。1つのプロジェクトの中では一貫してどちらかを使うことを推奨。

## 型を調べる：`type()` 関数

Python では `type(変数名)` を使うと、その変数の型を調べることができます。

```python
temperature = 23.8
location = "Lake Biwa"
print(type(temperature))
print(type(location))
```

```python
print(location + "の水温は" + str(temperature) + "℃です。")
```

# 1-3. 型変換とキーボード入力

## 型変換
- Pythonでは、ある型の値を別の型に変換することができる。  
- これを **型の変換（type conversion）** または **キャスト（casting）** と呼ぶ。
- 例:


```python
int("5")    # 文字列 "5" を整数 5 に変換
float("3.14")  # 文字列 "3.14" を小数 3.14 に変換
str(10)     # 整数 10 を文字列 "10" に変換
```

| 変換先の型 | 関数        | 説明                    | 例                                   |
| ----- | --------- | --------------------- | ----------------------------------- |
| 整数    | `int()`   | 小数や数値文字列を整数に変換        | `int("10") → 10`                    |
| 小数    | `float()` | 整数や数値文字列を小数に変換        | `float("3.5") → 3.5`                |
| 文字列   | `str()`   | 数値や論理値を文字列に変換         | `str(25) → "25"`                    |
| 論理値   | `bool()`  | 値を真偽（True / False）に変換 | `bool(0) → False`, `bool(1) → True` |


## キーボードからの入力
- input()という関数は、キーボードから入力された文字を文字列(str)として返す。
- キーボードから入力した値（input()）は文字列（str型）として扱われるため、計算に使うには数値型（intやfloat）へ変換する必要がある。

```python
print(input("入力された文字列を表示します:"))
```

```python
strin = input("摂氏の温度を入力してください:")
print(float(strin) + 273.15)
```

# 1-4. print()による出力とf-string整形

## 出力整形
- 出力整形とは、数値や文字列の見た目をきれいに整えることである。
- 見た目を整えることは一見「本質的ではない」ように思えるかもしれないが、実験結果や観測値を確認するのは人間である。人が視覚的に理解しやすい形でデータを出力できるようにすることには大きな意義がある。
- print()の場合、「小数点以下2桁まで表示したい」といったときf-string整形を用いる。

## f-string整形とは？
- f-string（フォーマット文字列）とは、文字列の前に f を付けて、中かっこ {} の中に変数や式を直接書ける構文。
- {} の中では 小数点以下の桁数や幅などを指定して整形表示ができる。

```python
temperature = 23.4567
print(f"気温は {temperature} 度です。")
print(f"気温は {temperature:.2f} 度です。")
```

| 書式指定       | 意味             | 例                    | 出力           |
| ---------- | -------------- | -------------------- | ------------ |
| `{x}`      | そのまま表示         | `{temperature}`      | 23.4567      |
| `{x:.1f}`  | 小数点以下1桁        | `{temperature:.1f}`  | 23.5         |
| `{x:.2f}`  | 小数点以下2桁        | `{temperature:.2f}`  | 23.46        |
| `{x:8.2f}` | 全体8桁・小数2桁（右寄せ） | `{temperature:8.2f}` | `'   23.46'` |


# 1-5. コメントアウトとエラーメッセージ

## なぜ型を意識する必要があるのか？

- **計算できるかどうかは型で決まる**  
    - たとえば、エクセルファイルで私たちの目に見える「5」という数字は、**文字列 `"5"`** の場合と **整数 `5`** の場合がある。  
    - 文字列のままでは計算に使えないため、一度 `int("5")` のように数値に変換する必要がある。

- 以下、# print(a+3) の行頭の # を除いた場合、どうなるか確認してみよう。
    - `#` はコメントアウトを表す記号で、この記号から始まる文章はプログラム実行時に無視される。
    - メモや説明文として利用できる他、エラーチェックの際に頻繁に用いる。

```python
# 例：文字列と整数の違い
a = "5"   # 文字列
b = 5     # 整数

# 以下はエラーになる
# print(a + 3)

# 正しくは、文字列を整数に変換
print(int(a) + 3)
```

- 与えられた値が指定した型の変換に適していない場合、たとえば文字列を `int()` や `float()` に変換しようとするとエラーが発生する。

```python
# OK
int("5")

# エラーになる例
int("five")   # ValueError: invalid literal for int()
```

- 赤文字で表示された ValueError はエラーメッセージの1つで、与えられた値が変換に適していないことを指摘する。
- これらのエラーメッセージは、**プログラムのどこが間違っているのかを知る重要な手がかり**。はじめのうちは英語で書かれているため戸惑うかもしれないが、すぐに意味がわからない場合はChatGPTなどに貼り付けて解釈を確認するとよい。

- print 関数は文字列を + で結合して表示することもできる。
- print 関数では数値の代入された変数を引数としてその数値を出力することもできる。
- しかしながら、文字列と一緒に出力することはできない。この場合、str() を使って数値を文字列に変換することで、文字列同士を結合して出力できる。

```python
print("今日の天気は"+"晴れです。")
```

```python
temperature = 23.5
print(temperature)
```

```python
temperature = 23.5
print("水温は " + temperature + " ℃です。")
```

```python
temperature = 23.5
print("水温は " + str(temperature) + " ℃です。")
```

# 1-6. 算術演算子と優先順位

- Python では数値を計算するために **算術演算子** を用いる。
- 四則演算のほか、累乗や余りの計算などもできる。多くはExcelで使用する演算子と同じ。
- 演算子には優先される順序があるため注意。

| 演算子  | 意味            | 例        | 結果  |
| ---- | ------------- | -------- | --- |
| `+`  | 加算            | `3 + 2`  | 5   |
| `-`  | 減算            | `5 - 1`  | 4   |
| `*`  | 乗算            | `2 * 4`  | 8   |
| `/`  | 除算            | `5 / 2`  | 2.5 |
| `//` | 整数除算（小数を切り捨て） | `5 // 2` | 2   |
| `%`  | 余り            | `5 % 2`  | 1   |
| `**` | 累乗            | `3 ** 2` | 9   |

- ⚠️ 演算子は ()  →  **  →  * / // %  →  + - の順で処理されるため注意。


```python
a = 10 + 2 * 3
b = (10 + 2) * 3
c = 10 + 2 ** 3
print(a, b, c)
```

# 1-7. import文と標準ライブラリ

- Python にはあらかじめ多くの **標準ライブラリ（道具箱）** が用意されている。
- 必要なときに import 文を使って取り出して使う。たとえば、数学関数をまとめた「math」モジュールを使うと、円周率 π や平方根などを扱うことができる。

```python
import math
# math.pi のピリオド . は "mathの中のpiを使う" という意味。
print(math.pi)       # 円周率 π
print(math.sqrt(16)) # 平方根
```
